### esercizio 2

```python
s1 = "Chiara" , 7.5 , 8 , 6.5 , 12.
s2 = "Lorenzo",6,4,7,9.


def media(studente):
    voti = studente[1:4]
    media = sum(voti)/len(voti)
    return media

def stato_assenze(studente):
    assenza = studente[4]
    if (assenza > 50):
        ammesso = "non valutabile"
    else: ammesso = "valutabile"
    return ammesso

def sufficenza(studente):
    if (media(studente)>= 6):
        suf = "sufficente"
    else: suf = "non sufficente"
    return suf

nome = s1[0]
media_valore = media(s1)
giudizio = sufficenza(s1)
assenze_valore = s1[4]
stato = stato_assenze(s1)

print(f"studente: {nome}, media: {media_valore:.2f}, giudizio: {giudizio}, assenze: {assenze_valore} ({stato})")

nome = s2[0]
media_valore = media(s2)
giudizio = sufficenza(s2)
assenze_valore = s2[4]
stato = stato_assenze(s2)

print(f"studente: {nome}, media: {media_valore:.2f}, giudizio: {giudizio}, assenze: {assenze_valore} ({stato})")
```

### esercizio 5

```python
testo = "Scrivere nella cella dell'esercizio 5 il programma che legge in ingresso una frase e conta, con un unico ciclo"
vocali = "aeiou"
consonanti = "bcdtfghjklmnpqrstvwxyz"

nvocali = 0
nconsonanti = 0
nnumeri = 0
nspeciali = 0

testo = testo.strip().lower()

for carattere in testo:
    if (carattere in vocali):
        nvocali+=1
    elif (carattere in consonanti):
        nconsonanti+=1
    elif (carattere.isdigit()):
        nnumeri+=1
    else: nspeciali+=1

percentualiVocal = nvocali / len(testo) *100

print(f"vocali:{nvocali}")
print(f"consonanti:{nconsonanti}")
print(f"numeri:{nnumeri}")
print(f"caratteri speciali:{nspeciali}")
print(f"percentuale vocali:{percentualiVocal:.3}")
```

### esercizio 9

```python
s = "A101;B204;C307;D410;E512"


cercato = input("Inserisci il codice da cercare: ").strip().lower()


testo = s + ";"
pos = 1


while len(testo) > 0:
   
    punto_virgola = testo.find(";")
    
    
    voce = testo[:punto_virgola]
    
    
    if voce.strip().lower() == cercato:
        print(f"Codice trovato alla posizione: {pos}")
        break
        
    
    testo = testo[punto_virgola + 1:]
    pos += 1
else:
    print("Codice assente")
```
