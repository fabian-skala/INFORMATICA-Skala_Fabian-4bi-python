## 3. Variabili, assegnamento e riferimenti

L'assegnamento non copia un valore dentro una cella, come si legge naturalmente in Java: lega un nome
a un oggetto. Due nomi possono quindi riferirsi allo stesso oggetto, e la funzione `id`, che restituisce
un identificatore univoco, permette di verificarlo. Con i tipi immutabili di questo modulo la differenza
non produce ancora effetti visibili; diventerà decisiva in M3, con liste e dizionari.


```python
voto_scritto = 8
voto_registro = voto_scritto

print(id(voto_scritto) == id(voto_registro))   # True: un solo oggetto, due nomi

voto_registro = 7                              # il nome passa a un altro oggetto
print(voto_scritto, voto_registro)
print(id(voto_scritto) == id(voto_registro))   # False
```

```python
# Assegnamento multiplo: la parte destra viene valutata per intero prima di assegnare.
primo, secondo = 3, 9
print(primo, secondo)

primo, secondo = secondo, primo    # scambio senza variabile d'appoggio
print(primo, secondo)

# Assegnamento composto: identico a Java. Non esistono ++ e --.
contatore = 0
contatore += 1
contatore += 1
print(contatore)

totale = 100
totale -= 15
totale *= 2
print(totale)
```

## 4. Operatori aritmetici e le tre divisioni

Somma, sottrazione e moltiplicazione si comportano come in Java. Le sorprese arrivano con la divisione.
La barra singola `/` esegue sempre la divisione reale e restituisce sempre un `float`, anche quando gli
operandi sono interi e il risultato è esatto. La doppia barra `//` esegue la divisione intera troncando
verso il basso. L'operatore `%` restituisce il resto ed è coerente con `//`. L'operatore `**` eleva a
potenza e rende inutile `Math.pow`.


```python
minuti_totali = 197

ore = minuti_totali // 60      # divisione intera: quante ore intere
minuti = minuti_totali % 60    # resto: i minuti che avanzano
frazione = minuti_totali / 60  # divisione reale: le ore con la virgola

print(ore, "ore e", minuti, "minuti")
print("in ore decimali:", frazione)

print(10 / 2)      # 5.0 e non 5: il tipo del risultato è float
print(10 // 3)
print(10 % 3)
print(2 ** 10)
```

```python
# Con i numeri negativi Python e Java non coincidono: Java tronca verso lo zero,
# Python arrotonda verso meno infinito e dà al resto il segno del divisore.
print(-7 // 2)     # -4   in Java -7 / 2 vale -3
print(-7 % 2)      #  1   in Java -7 % 2 vale -1
print(7 // -2)
print(7 % -2)

# la relazione fra // e % vale comunque sempre
a = -7
b = 2
print(a == (a // b) * b + a % b)

# l'elevamento a potenza associa a destra
print(2 ** 3 ** 2)     # 512, cioè 2 ** 9
print((2 ** 3) ** 2)   # 64
```

## 5. Confronti, identità e operatori logici

In Java `==` fra due oggetti confronta i riferimenti, ed è per questo che sulle stringhe si usava
`equals`. In Python `==` confronta il contenuto, mentre il confronto dei riferimenti si scrive `is`.
La regola operativa del corso non ammette eccezioni: `==` per i valori, `is None` per verificare
l'assenza di un valore, mai `is` su numeri e stringhe. Il confronto concatenato, che Java non ha,
permette di scrivere una fascia di valori nella forma in cui la si legge.


```python
primo = "informatica"
secondo = "informa" + "tica"

print(primo == secondo)    # True: stesso contenuto, nessun equals necessario
print(primo != secondo)

esito = None
print(esito is None)       # True: la forma corretta per l'assenza di valore

voto = 7
print(6 <= voto <= 8)      # confronto concatenato: voto viene valutato una volta sola
print(6 <= voto and voto <= 8)
```

Gli operatori logici si scrivono con parole inglesi: `and`, `or`, `not`. Il comportamento è quello di
`&&`, `||` e `!`, compresa la valutazione pigra. Python estende inoltre la nozione di verità a tutti gli
oggetti: sono considerati falsi lo zero in tutte le sue forme, la stringa vuota e `None`, mentre tutto
il resto è considerato vero.


```python
media = 5.5
ore_assenza = 12

promosso = media >= 6 and ore_assenza <= 50
bocciato = media < 6 or ore_assenza > 50
print(promosso, bocciato, not promosso)

nota = ""
if not nota:
    print("nessuna nota registrata")   # la stringa vuota vale False

print(bool(0), bool(0.0), bool(""), bool(None))
print(bool(-1), bool("0"), bool(" "))
```

L'unica conversione automatica di Python riguarda i numeri: in un'espressione che mescola `int` e
`float` l'intero viene promosso. Fuori dai numeri non esiste alcuna conversione implicita, e questo è il
punto in cui Python si distingue nettamente da Java, che accetta `"Media: " + 7.5` e converte in
silenzio. Le conversioni esplicite si ottengono chiamando il nome del tipo come se fosse una funzione.


```python
print(3 + 0.5)               # l'intero viene promosso a float
print(type(3 + 0.5))
print(True + True)           # bool è un caso particolare di int: True vale 1

print(7 // 2, type(7 // 2))
print(7 / 2, type(7 / 2))
```

```python
# nel programma reale:  lettura_sensore = input("Temperatura rilevata: ")
lettura_sensore = "23.7"

temperatura = float(lettura_sensore)
print(temperatura + 1.0)

print(int(temperatura))      # 23: int tronca, non arrotonda
print(round(temperatura))    # 24: round arrotonda
print(int(-7.9))             # -7: il troncamento va verso lo zero

print(int("42"), float("42"), str(42))
print(int(float("7.5")))     # passaggio obbligato per le stringhe con decimali
```

## 7. Uscita e f-string

La funzione `print` separa gli argomenti con uno spazio e va a capo alla fine; il separatore e il
terminatore si cambiano con `sep` ed `end`. La formattazione si fa con le f-string, cioè stringhe
precedute dalla lettera `f` in cui qualsiasi espressione fra parentesi graffe viene valutata e sostituita
dal suo valore. Dopo i due punti si scrive lo specificatore di formato, che governa precisione,
allineamento e ampiezza: è l'equivalente compatto di `String.format` e di `printf`.


```python
print("giorno", "mese", "anno")
print("2026", "10", "12", sep="-")

print("Caricamento", end="")     # non va a capo
print("... completato")

print()                          # riga vuota
print("fine")
```

```python
studente = "Chiara"
scritto = 7.5
orale = 8.0
media = (scritto + orale) / 2

print(f"{studente}: scritto {scritto}, orale {orale}")
print(f"Media: {media}")
print(f"Media a due decimali: {media:.2f}")
print(f"Media in percentuale: {media / 10:.1%}")
print(f"Somma dei due voti: {scritto + orale}")
print(f"Le graffe letterali si raddoppiano: {{media}}")
```

```python
# Incolonnamento: < a sinistra, > a destra, ^ centrato, seguiti dall'ampiezza.

print(f"|{'Studente':<12}|{'Media':>8}|")
print("-" * 23)
print(f"|{'Chiara':<12}|{7.75:>8.2f}|")
print(f"|{'Alessandro':<12}|{6.5:>8.2f}|")
print(f"|{'Totale':<12}|{14.25:>8.2f}|")
print(f"{'fine tabella':^23}")
```

## 8. Selezione e ruolo dell'indentazione

In Python le parentesi graffe non ci sono e l'indentazione è sintassi: le istruzioni di un blocco si
riconoscono dal rientro, e la riga che apre il blocco termina sempre con i due punti. La convenzione è
di quattro spazi per livello. La parola chiave `elif` sostituisce l'annidamento `else if` di Java, i rami
vengono provati nell'ordine in cui compaiono e si esegue il primo la cui condizione risulta vera.


```python
media = 7.2
assenze = 18

if assenze > 50:
    esito = "non valutabile"
elif media >= 8:
    esito = "ottimo"
elif media >= 6:
    esito = "sufficiente"
else:
    esito = "insufficiente"

print(f"media {media}, assenze {assenze} ore, esito: {esito}")
```

```python
voto = 5

# espressione condizionale: prima il valore se vero, poi la condizione, poi l'alternativa
esito = "promosso" if voto >= 6 else "respinto"
print(esito)
print(f"Lo studente è {'promosso' if voto >= 6 else 'respinto'}")

# l'indentazione decide che cosa sta dentro il blocco e che cosa sta fuori
if voto >= 6:
    print("dentro il blocco: compare solo se la condizione è vera")
print("fuori dal blocco: compare sempre")
```

## 9. Il ciclo while

Il `while` ripete un blocco finché la condizione resta vera e si comporta esattamente come in Java. È la
struttura da usare quando il numero di ripetizioni non è noto in anticipo. Il costrutto `do ... while`
non esiste: quando serve eseguire il corpo almeno una volta si scrive `while True:` con un `break` nel
punto opportuno.


```python
# somma dei primi n interi con il ciclo a condizione
n = 10
i = 1
somma = 0
while i <= n:
    somma += i
    i += 1
print(f"somma dei primi {n} interi: {somma}")

# conto alla rovescia
secondi = 5
while secondi > 0:
    print(secondi, end=" ")
    secondi -= 1
print("via")
```

### Una sequenza di dati senza liste

Il programma reale leggerebbe i voti uno alla volta con `input`, ripetendo la lettura in fondo al corpo
del ciclo. Qui la stessa sequenza è scritta in una sola stringa, con il punto e virgola a separare i
valori, e si scorre con `find` e slicing. Lo schema è più laborioso di quanto sarà necessario da M3 in
poi, quando il metodo `split` restituirà direttamente una lista di valori, ma usa soltanto quello che
sappiamo ora ed è un ottimo esercizio di elaborazione di stringhe.


```python
voti = "7;5.5;9;6;8"

posizione = 0
somma = 0.0
quanti = 0

while posizione < len(voti):
    fine = voti.find(";", posizione)   # cerca il separatore a partire da posizione
    if fine == -1:                     # nessun separatore: siamo sull'ultimo valore
        fine = len(voti)
    voto = float(voti[posizione:fine])
    somma += voto
    quanti += 1
    posizione = fine + 1               # si riparte dopo il separatore

print(f"voti letti: {quanti}")
print(f"media: {somma / quanti:.2f}")
```

## 10. Il ciclo for e la funzione range

Il `for` di Python non è il `for` a contatore di Java: scorre gli elementi di una sequenza, uno dopo
l'altro, e corrisponde al for-each. Il ciclo a contatore si ottiene scorrendo la sequenza di interi
prodotta da `range`, che genera i numeri con l'estremo inferiore incluso e quello superiore escluso.


```python
for i in range(5):
    print(i, end=" ")
print()

for i in range(1, 6):
    print(i, end=" ")
print()

for i in range(0, 21, 5):
    print(i, end=" ")
print()

for i in range(10, 0, -2):
    print(i, end=" ")
print()

print(range(1000000))   # range non costruisce i numeri: li produce uno alla volta
```

```python
# la tabellina: l'equivalente diretto del for a contatore di Java
tabella = 7
for i in range(1, 11):
    print(f"{tabella} x {i:2d} = {tabella * i:3d}")
```

Il `for` scorre altrettanto bene i caratteri di una stringa, senza bisogno di alcun indice, e questa è la
forma da preferire quando l'indice non serve. L'operatore `in`, applicato a una stringa, verifica se una
sottostringa vi compare: `carattere in "aeiou"` sostituisce la catena di confronti che servirebbe in Java.


```python
frase = "Nel laboratorio di informatica si programma in Python"

vocali = 0
consonanti = 0
for carattere in frase.lower():
    if carattere in "aeiou":
        vocali += 1
    elif carattere.isalpha():
        consonanti += 1

print(f"vocali: {vocali}, consonanti: {consonanti}")
print(f"caratteri in tutto: {len(frase)}")
```

## 11. break, continue e la clausola else dei cicli

Le istruzioni `break` e `continue` hanno lo stesso significato che in Java. La clausola `else` di un
ciclo, che Java non ha, viene eseguita quando il ciclo termina in modo naturale e viene saltata quando è
un `break` a interromperlo: va letta come «se non c'è stato alcun break». È la sede sintattica del caso
«cercato e non trovato», che in Java richiedeva una variabile booleana di appoggio.


```python
# verifica di primalità su un intervallo di numeri
for n in range(95, 102):
    divisore = 2
    while divisore * divisore <= n:
        if n % divisore == 0:
            print(f"{n:3d} non è primo: divisibile per {divisore}")
            break
        divisore += 1
    else:
        print(f"{n:3d} è primo")
```

```python
# continue: le letture pari a -1 segnalano un guasto del sensore e vanno ignorate
letture = "18.4;-1;21.9;-1;20.2"

posizione = 0
somma = 0.0
valide = 0
scartate = 0

while posizione < len(letture):
    fine = letture.find(";", posizione)
    if fine == -1:
        fine = len(letture)
    valore = float(letture[posizione:fine])
    posizione = fine + 1               # aggiornato prima del continue, altrimenti il ciclo non finisce
    if valore < 0:
        scartate += 1
        continue
    somma += valore
    valide += 1

print(f"letture valide: {valide}, scartate: {scartate}")
print(f"media delle valide: {somma / valide:.2f}")
```

## 12. Le stringhe come sequenze

Una stringa si misura con `len`, si indicizza con le parentesi quadre a partire da zero e si affetta con
lo slicing. La novità rispetto a Java sono gli indici negativi, che contano dalla fine: `parola[-1]`
sostituisce `parola.charAt(parola.length() - 1)`.

## 13. Algoritmi già noti in Java, riscritti in Python

Gli algoritmi che seguono sono gli stessi scritti in Java al terzo anno. Riscriverli serve a verificare
che il ragionamento si trasferisca senza perdite e a misurare quanta parte del codice Java fosse
impalcatura del linguaggio piuttosto che sostanza dell'algoritmo.


```python
# Massimo, minimo e media di una sequenza di voti.
voti = "7;5.5;9;6;8"

posizione = 0
massimo = 0.0
minimo = 0.0
somma = 0.0
quanti = 0

while posizione < len(voti):
    fine = voti.find(";", posizione)
    if fine == -1:
        fine = len(voti)
    voto = float(voti[posizione:fine])
    posizione = fine + 1

    if quanti == 0:            # il primo valore letto inizializza massimo e minimo
        massimo = voto
        minimo = voto
    else:
        if voto > massimo:
            massimo = voto
        if voto < minimo:
            minimo = voto

    somma += voto
    quanti += 1

print("-" * 30)
print(f"voti letti: {quanti}")
print(f"massimo: {massimo:.2f}")
print(f"minimo:  {minimo:.2f}")
print(f"media:   {somma / quanti:.2f}")
```

```python
# Massimo comun divisore con l'algoritmo di Euclide.
a = 1071
b = 462

x = abs(a)
y = abs(b)
while y != 0:
    x, y = y, x % y      # l'assegnamento multiplo elimina la variabile d'appoggio

print(f"MCD({a}, {b}) = {x}")
```

```python
# Somma delle cifre e conversione in binario.
n = 156

resto_lavoro = n
somma_cifre = 0
while resto_lavoro > 0:
    somma_cifre += resto_lavoro % 10
    resto_lavoro = resto_lavoro // 10
print(f"somma delle cifre di {n}: {somma_cifre}")

resto_lavoro = n
binario = ""
while resto_lavoro > 0:
    binario = str(resto_lavoro % 2) + binario   # si costruisce da destra a sinistra
    resto_lavoro = resto_lavoro // 2
if binario == "":
    binario = "0"

print(f"{n} in binario è {binario}")
print(f"verifica con la funzione predefinita: {bin(n)[2:]}")
```

```python
# Riconoscimento di una frase palindroma, nelle due versioni.
frase = "I topi non avevano nipoti"

ripulita = ""
for carattere in frase.lower():
    if carattere.isalpha():
        ripulita += carattere
print("testo normalizzato:", ripulita)

print("con lo slicing:", ripulita == ripulita[::-1])

# versione con i due indici che si avvicinano, identica a quella scritta in Java
sinistra = 0
destra = len(ripulita) - 1
palindroma = True
while sinistra < destra:
    if ripulita[sinistra] != ripulita[destra]:
        palindroma = False
        break
    sinistra += 1
    destra -= 1
print("con i due indici:", palindroma)
```

```python
# Ricerca sequenziale voce per voce, come sull'array Java.
elenco = "A101;B204;C307;D410;E512"
cercato = "C307"

posizione = 0
indice_voce = 0
trovato = -1
while posizione < len(elenco):
    fine = elenco.find(";", posizione)
    if fine == -1:
        fine = len(elenco)
    voce = elenco[posizione:fine]
    if voce == cercato:
        trovato = indice_voce
        break
    posizione = fine + 1
    indice_voce += 1

if trovato == -1:
    print(f"{cercato} non è presente nell'elenco")
else:
    print(f"{cercato} è la voce numero {trovato + 1}")

# lo stesso lavoro fatto dal metodo find, che restituisce la posizione nella stringa
print("posizione nella stringa:", elenco.find(cercato))
print("codice assente:", elenco.find("Z999"))
```

### Tradurre senza array e senza `def`

Le due celle che seguono sono il punto in cui la distanza fra i due linguaggi si vede meglio. Nella prima,
la conversione da binario a decimale scorre i caratteri della stringa senza indice e senza `charAt`, e la
cifra si ottiene con `int(carattere)` invece che con la sottrazione del codice del carattere zero.

Nella seconda, un metodo Java che riceveva un array e restituiva una posizione diventa codice scritto per
esteso: la sequenza sta in una stringa separata dal punto e virgola, la scansione avanza con il secondo
argomento di `find`, che indica da quale posizione riprendere la ricerca, e il risultato resta nelle
variabili invece di essere restituito. Il codice è più lungo della versione Java, non più corto: gli array
tornano come liste in M3 e i metodi come funzioni in M4, e allora lo stesso programma si accorcerà.

```python
# Da binario a decimale: lo stesso schema scritto in Java su un array di caratteri.
binario = "11010"          # nel programma reale:  binario = input("Sequenza di zeri e uni: ").strip()

valida = len(binario) > 0
for carattere in binario:
    if carattere not in "01":
        valida = False

if not valida:
    print("La sequenza non è un numero binario.")
else:
    valore = 0
    for carattere in binario:
        valore = valore * 2 + int(carattere)
    print(f"{binario} in base due vale {valore} in base dieci")
    print(f"verifica con la funzione predefinita: {int(binario, 2)}")
```

```python
# In Java era il metodo "int indiceDelMassimo(double[] v)": qui niente array e niente def.
letture = "19.0;19.6;20.4;22.3;21.1;20.8"

inizio = 0
indice = 0
posizione = 0
massimo = float(letture[0:letture.find(";")])

while inizio < len(letture):
    fine = letture.find(";", inizio)
    if fine == -1:              # ultima voce: non è seguita dal separatore
        fine = len(letture)
    valore = float(letture[inizio:fine])
    if valore > massimo:
        massimo = valore
        posizione = indice
    inizio = fine + 1           # si riparte dal carattere dopo il separatore
    indice += 1

print(f"massimo {massimo:.1f} in posizione {posizione} su {indice} letture")
```

## 14. Errori frequenti

Gli errori che seguono non vengono eseguiti, perché interromperebbero il notebook. Il messaggio
dell'interprete va letto dal basso verso l'alto: l'ultima riga contiene il nome dell'errore e la
descrizione, le righe sopra indicano la posizione nella forma `Cell In[3], line 2`, con una freccia sul
punto sospetto. Il numero fra parentesi quadre è quello dell'esecuzione, non quello della cella nel
documento, e cambia ogni volta che la cella viene rieseguita.

Il rientro mancante dopo i due punti produce `IndentationError: expected an indented block after 'if'
statement on line 1`; una riga rientrata senza che nulla abbia aperto un blocco produce
`IndentationError: unexpected indent`; il livello di rientro che non corrisponde a nessun blocco aperto
produce `unindent does not match any outer indentation level`, che nasce quasi sempre da tabulazioni
mescolate a spazi.

Il risultato di `input` usato senza conversione produce `TypeError: can only concatenate str (not
"int") to str`, e il difetto sta nella riga precedente a quella segnalata. La conversione di una
stringa che non contiene un intero produce `ValueError: invalid literal for int() with base 10: '7.5'`.
Il segno di assegnamento al posto del confronto produce `SyntaxError: invalid syntax. Maybe you meant
'==' or ':=' instead of '='?`. Un nome scritto con la maiuscola sbagliata produce `NameError: name
'Totale' is not defined`, con il suggerimento del nome corretto. Un indice fuori intervallo produce
`IndexError: string index out of range`.

Più insidiosi sono gli errori che non producono alcun messaggio: la cella che segue ne mostra tre.

## 15. Correzione guidata: la procedura in cinque passi

Correggere un programma che non funziona non è questione di intuito: è una procedura, e conviene eseguirla
sempre nello stesso ordine invece di cambiare righe a caso. Si legge l'ultima riga del messaggio e la si
traduce in italiano; si individua la riga segnalata ricordando che il difetto sta spesso in quella
precedente; si stampano valore e tipo delle variabili coinvolte con `print` e `type`; si corregge una cosa
sola e si riesegue, perché un secondo errore nascosto dietro il primo è la norma; e quando non compare
alcun messaggio ma il risultato è sbagliato si controllano i quattro sospetti abituali, cioè la barra
singola al posto della doppia, il metodo di stringa chiamato senza assegnarne il risultato, la
concatenazione scambiata per una somma e la condizione del `while` che non cambia mai.

La prima cella mostra il terzo passo, che è quello risolutivo nella maggior parte dei casi. La seconda
riporta un programma difettoso nella forma di commento, con i suoi tre difetti dichiarati, e subito sotto
la versione riparata che viene effettivamente eseguita.

```python
# Versione difettosa, lasciata come commento perché non arriverebbe in fondo:
#
#   voti = "7;6;9;8"
#   totale = 0
#   quanti = 0
#   inizio = 0
#   while inizio < len(voti):
#       fine = voti.find(";", inizio)
#       totale += voti[inizio:fine]
#       quanti += 1
#       inizio = fine + 1
#   print("media:", totale / quanti)
#
# Tre difetti: la somma concatena una stringa a un intero e solleva TypeError;
# sull'ultima voce find restituisce -1 e la voce viene troncata; con fine a -1
# l'indice torna a zero e il ciclo non termina mai.

voti = "7;6;9;8"
totale = 0
quanti = 0
inizio = 0

while inizio < len(voti):
    fine = voti.find(";", inizio)
    if fine == -1:
        fine = len(voti)
    totale += int(voti[inizio:fine])
    quanti += 1
    inizio = fine + 1

print(f"voti letti: {quanti}, media: {totale / quanti:.2f}")
```

## Che cosa portare via da questo notebook

Il tipo appartiene all'oggetto e non al nome, e viene verificato durante l'esecuzione: un errore di tipo
in un ramo mai percorso non si manifesta finché quel ramo non viene percorso. La barra singola restituisce
sempre un `float`, la doppia barra tronca verso il basso e il resto segue il segno del divisore. La
funzione `input` restituisce sempre una stringa e la conversione è a carico di chi scrive il programma;
l'uscita si formatta con le f-string, che incorporano espressioni e specificatori di allineamento.

L'indentazione definisce i blocchi e la riga che ne apre uno termina con i due punti. Il `for` scorre
sequenze e diventa un ciclo a contatore quando lo si abbina a `range`; la clausola `else` del ciclo
esprime il caso «cercato e non trovato» senza la variabile booleana che serviva in Java. Le stringhe
sono sequenze immutabili: si indicizzano anche dalla fine, si affettano con lo slicing e i loro metodi
restituiscono sempre una stringa nuova.

Il modulo M3 sostituirà la scansione con `find` e slicing usata qui per le sequenze di dati: liste,
tuple, dizionari e insiemi rendono immediato quello che in questo notebook ha richiesto un ciclo scritto
a mano. Il modulo M4 introdurrà `def` e permetterà di dare un nome ai frammenti di codice che qui sono
stati ripetuti.
