```python
riga_registro = "  2026-11-20;4BI;laboratorio;2  "

# 1. Pulizia della riga da spazi bianchi iniziali e finali
pulita = riga_registro.strip()
print(f"Riga pulita: {pulita}")

# 2. Estrazione dei quattro campi usando il separatore ';'
campi = pulita.split(";")

data = campi[0]
classe = campi[1]
attivita = campi[2]
ore = campi[3]

# 3. Stampa del riepilogo formattato
print("\n--- RIEPILOGO REGISTRO ---")
print(f"Data: {data}")
print(f"Classe: {classe}")
print(f"Attività: {attivita}")
print(f"Ore: {ore}")
```
