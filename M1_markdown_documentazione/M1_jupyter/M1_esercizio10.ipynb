# Consegnare un esercizio del corso

**Destinatario:** i compagni di classe che consegnano per la prima volta.

**Scopo:** portare un file dal proprio computer al repository remoto in sei passi. Senza questa procedura il docente non vede il lavoro e non può valutarlo.

## Passo 1: modifica il file

Lavora solo dentro la cartella del modulo, per esempio `M1_markdown/`. Se salvi altrove, `git` non trova il file nel posto previsto e la consegna resta incompleta.

```bash
code M1_markdown/postazione.md
```

## Passo 2: controlla lo stato

Il comando mostra quali file sono cambiati. Se lo salti, rischi di consegnare file che non volevi o di dimenticarne uno.

```bash
git status
```

````python
passi = 6
print("Numero di passi:", passi)
````

## Passo 3: prepara i file

Segnala a Git quali file entrano nel prossimo commit. Se lo salti, il commit risulta vuoto e il lavoro non viene registrato.

```bash
git add M1_markdown/postazione.md
```

## Passo 4: registra il commit

Il messaggio dice che cosa contiene la consegna ed è la voce che il docente legge nella cronologia. Un messaggio generico non permette di ritrovare l'esercizio.

```bash
git commit -m "Aggiunge la scheda della postazione (M1 es. 1)"
```

````python
minuti_modifica = 20
minuti_git = 5
minuti_totali = minuti_modifica + minuti_git
print("Minuti stimati:", minuti_totali)
````

## Passo 5: invia il commit

Il commit esiste solo sul tuo computer finché non lo invii. Se salti questo passo, il repository remoto non cambia.

```bash
git push
```

## Passo 6: controlla su GitHub

Apri la pagina del repository nel browser e guarda il file reso. Se salti il controllo, errori come tabelle non riconosciute o immagini mancanti arrivano al docente.

```bash
git log --oneline -1
```

````python
file_modificati = 1
file_nuovi = 0
print("File nel commit:", file_modificati + file_nuovi)
````

## Riepilogo dei comandi

| Comando | Effetto |
| :------ | :------ |
| `git status` | elenca i file cambiati |
| `git add` | prepara i file per il commit |
| `git commit -m` | registra le modifiche in locale |
| `git push` | invia i commit al repository remoto |
| `git log --oneline -1` | mostra l'ultimo commit registrato |

> Prima di consegnare, apri l'anteprima del documento e controlla che ogni elemento sia reso come previsto. Un documento mai guardato è un documento non verificato.
